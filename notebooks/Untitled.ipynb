```python
import requests
import json

BASE_URL = "http://localhost:1143"

def execute_query(sql: str):
    try:
        response = requests.get(
            f"{BASE_URL}/query",
            params={
                "query": sql,
                "format": "JSON"
            },
            timeout=30
        )

        response.raise_for_status()

        data = response.json()

        if data.get("error"):
            print(f"SQL Error: {data['error']}")
            return None

        return data

    except requests.exceptions.Timeout:
        print("Request timed out")
    except requests.exceptions.ConnectionError:
        print("Could not connect to server")
    except requests.exceptions.HTTPError as e:
        print(f"HTTP Error: {e}")
    except Exception as e:
        print(f"Unexpected error: {e}")

    return None


if __name__ == "__main__":
    # 1. Create the database
    print("Creating database USERS...")
    db_result = execute_query("CREATE DATABASE USERS")
    if db_result:
        print(json.dumps(db_result, indent=2))

    # 2. Create a table using the File storage engine
    # In Mnemosyne, we select the storage engine at table-creation time.
    print("\nList databases...")
    table_query = """SHOW DATABASES;"""
    table_result = execute_query(table_query)
    if table_result:
        print(json.dumps(table_result, indent=2))
```
